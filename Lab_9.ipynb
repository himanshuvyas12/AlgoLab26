# Shortest Path Using Dijkstra Algorithm in Python
Problem Description:
Given a weighted graph with non negative edge weights. Find the shortest distance from a source vertex to all other vertices using Dijkstra algorithm

#Theory
Dijkstra algorithm finds shortest distance from one source node to all other nodes in a weighted graph with non negative weights. Dictionary structure stores nodes as keys and neighbor nodes with weights as values.

##Algorithm Steps

1. Represent graph using dictionary.
2. Set distance of every node as infinity.
3. Set distance of source node as 0.
4. Select node with smallest distance among unvisited nodes.
5. Update distance of neighbor nodes.
6. Repeat until all nodes get visited.

```python
def dijkstra(graph, start):

    distances = {}
    predecessors = {} # To store the path
    unvisited_nodes = set(graph.keys()) # Keep track of unvisited nodes

    for node in graph:
        distances[node] = float('inf')
        predecessors[node] = None # Initialize predecessors

    distances[start] = 0

    while unvisited_nodes:

        min_node = None
        min_distance = float('inf')

        # Find the unvisited node with the smallest distance
        for node in unvisited_nodes:
            if distances[node] < min_distance:
                min_distance = distances[node]
                min_node = node

        if min_node is None: # If no reachable unvisited node is found, break
            break

        unvisited_nodes.remove(min_node)

        # If the min_distance is infinity, all remaining nodes are unreachable from the start node
        if min_distance == float('inf'):
            break

        for neighbor, weight in graph[min_node].items():
            new_distance = distances[min_node] + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                predecessors[neighbor] = min_node # Update predecessor

    return distances, predecessors # Return both distances and predecessors
```

```python
graph = {
'A': {'B': 4, 'C': 2},
'B': {'A': 4, 'C': 5, 'D': 10},
'C': {'A': 2, 'B': 5, 'D': 3},
'D': {'B': 10, 'C': 3}
}

result = dijkstra(graph, 'A')

for node in result:
    print("Distance from A to", node, "=", result[node])
```

```python
def get_shortest_path(predecessors, start, target):
    path = []
    current = target
    while current is not None:
        path.insert(0, current)
        if current == start:
            break
        current = predecessors.get(current)

    if not path or path[0] != start:
        return None
    return path

graph = {
    'A': {'B': 4, 'C': 2},
    'B': {'A': 4, 'C': 5, 'D': 10},
    'C': {'A': 2, 'B': 5, 'D': 3},
    'D': {'B': 10, 'C': 3},
    'E': {'F': 1},
    'F': {'E': 1, 'G': 2},
    'G': {}
}

for node in list(graph.keys()):
    for neighbor in graph[node].keys():
        if neighbor not in graph:
            graph[neighbor] = {}

start_node = 'A'
distances, predecessors = dijkstra(graph=graph, start=start_node)

print(f"Distances from {start_node}:")
for node in distances:
    print(f"Distance from {start_node} to {node} = {distances[node]}")

print(f"\nShortest Paths from {start_node}:")
for node in sorted(distances.keys()):
    path = get_shortest_path(predecessors, start_node, node)
    if path:
        total_distance = distances[node]
        print(f"Path from {start_node} to {node}: {' -> '.join(path)} (Total Distance: {total_distance})")
    else:
        if distances[node] == float('inf'):
            print(f"Path from {start_node} to {node}: No path (unreachable)")
        else:
            print(f"Path from {start_node} to {node}: Error in path reconstruction (Distance: {distances[node]})")
```
