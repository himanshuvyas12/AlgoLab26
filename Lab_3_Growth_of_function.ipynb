# 1. Growth Of Function
Time Complexity/Order of Growth defines the amount of time taken by any program with respect to the size of the input. Time Complexity specifies how the program would behave as the order of size of input is increased. So, Time Complexity is just a function of size of its input. We will go through some of basic and most common time complexities such as:

- *Constant Time Complexity O(1) : constant running time*
- *Linear Time Complexity O(n) : linear running time*
- *Logarithmic Time Complexity O(log n) : logarithmic running time*
- *Log-Linear Time Complexity O(n log n) : log-­linear running time*
- *Polynomial Time Complexity O(n^c) : polynomial running time (c is a constant)*
- *Exponential Time Complexity O(c^n) : exponential running time (c is a constant being raised to a power based on size of input)*

## Constant time Complexity:O(1)
An algorithm is said to have constant time with order O (1) when it is not dependent on the input size n. Irrespective of the input size n, the runtime will always be the same. See example below

```python
# Load the Python libraries
import matplotlib.pyplot as plt
```

```python
# Prob_1

def add_two_numbers(a, b):
    return a + b

n_values = list(range(1, 101))
time_values = []

for n in n_values:
    add_two_numbers(n, n + 1)
    time_values.append(1)

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(1) Constant Time Complexity for Addition")
plt.show()
```

#Linear Time Complexity O(n) : linear running time
- *An algorithm is said to have a linear time complexity when the running time increases linearly with the length of the input. When the function involves checking all the values in input data, with this order O(n).*

```python
def linear_time_sum(numbers):
    total = 0
    for value in numbers:
        total += value
    return total

n_values = list(range(1, 101))
time_values = []

for n in n_values:
    data = list(range(n))
    linear_time_sum(data)
    time_values.append(n)

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(n) Linear Time Complexity")
plt.show()
```

#Logarithmic Time Complexity O(log n) : logarithmic running time
- *An algorithm is said to have a logarithmic time complexity when it reduces the size of the input data in each step. This indicates that the number of operations is not the same as the input size. The number of operations gets reduced as the input size increases. Algorithms are found in binary tree or binary search functions. This involves the search of a given value in an array by splitting the array into two and starting searching in one split. This ensures the operation is not done on every element of the data.*

```python
#Prob_4 Take Suitable example and write function that runs in logarithic time. plot graph as well similar to above example.

import math
import matplotlib.pyplot as plt

# Function to simulate logarithmic time complexity (Binary Search)
def binary_search_simulated_time(n):
    # In a true binary search, the number of comparisons is log base 2 of n.
    # We'll use math.log2(n) for a more accurate representation of binary search steps.
    if n <= 1: # Logarithm of 1 is 0, but for practical purposes, it takes at least 1 step.
        return 1
    return math.log2(n)

# Define a range of input sizes
n_values = list(range(1, 1001))

# Calculate simulated time values
time_values = []
for n in n_values:
    time_values.append(binary_search_simulated_time(n))

# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time (log n)")
plt.title("O(log n) Logarithmic Time Complexity (Binary Search Simulation)")
plt.grid(True)
plt.show()
```

#*Log-Linear Time Complexity O(n log n) : log-­linear running time*


* When we call a Logarithmic Time Algorithm inside a loop, it would
result into a Log-Linear Time Complexity program.




```python
#Prob_5 Take Suitable example and write function that runs in log-linear time. plot the graph as well, similar to above example.

import math
import matplotlib.pyplot as plt

# Function to simulate log-linear time complexity (e.g., Merge Sort)
def merge_sort_simulated_time(n):
    if n <= 1:
        return 1
    # Merge sort performs n * log(n) operations
    return n * math.log2(n)

# Define a range of input sizes
n_values = list(range(1, 1001))

# Calculate simulated time values
time_values = []
for n in n_values:
    time_values.append(merge_sort_simulated_time(n))

# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time (n log n)")
plt.title("O(n log n) Log-Linear Time Complexity (Merge Sort Simulation)")
plt.grid(True)
plt.show()
```

### Polynomial Time Complexity O(n^c) : polynomial running time (c is a constant)

* When the computation time increases as function of N raised to some power, N being the size of input. Such a code has Polynomial Time complexity.




```python
# Prob_6 Take Suitable example and write function that runs in log-linear time. plot the graph as well, similar to above example
import matplotlib.pyplot as plt

# Function to simulate polynomial time complexity (e.g., O(n^2) for nested loops)
def polynomial_simulated_time(n, c=2):
    # For O(n^c), we can simply return n raised to the power of c.
    # Using c=2 for O(n^2) as a common example.
    return n**c

# Define a range of input sizes
n_values = list(range(1, 101)) # Keeping n smaller for polynomial to visualize better

# Calculate simulated time values
time_values = []
for n in n_values:
    time_values.append(polynomial_simulated_time(n))

# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time (n^2)")
plt.title("O(n^2) Polynomial Time Complexity (Nested Loops Simulation)")
plt.grid(True)
plt.show()
```

#Exponential Time Complexity O(c^n) : exponential running time (c is a constant being raised to a power based on size of input)

* When the computation time of a code increases as function of X^N, N being the size of input. Such a code has Polynomial Time Complexity.




```python
# Prob_7 Take Suitable example and write function that runs in log-linear time. plot the graph as well, similar to above example

import matplotlib.pyplot as plt

# Function to simulate exponential time complexity (e.g., O(c^n))
def exponential_simulated_time(n, c=2):
    # For O(c^n), we can simply return c raised to the power of n.
    # Using c=2 as a common example.
    return c**n

# Define a range of input sizes (keep n small for exponential growth)
n_values = list(range(1, 15)) # 1 to 14, as it grows very fast

# Calculate simulated time values
time_values = []
for n in n_values:
    time_values.append(exponential_simulated_time(n))

# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time (c^n)")
plt.title("O(c^n) Exponential Time Complexity Simulation")
plt.grid(True)
plt.show()
```

```python
#Prob_8 plot graph similar to above figure

import math
import matplotlib.pyplot as plt
import numpy as np

# --- Define time complexity functions ---

def O_1(n): # Constant Time
    return 1

def O_log_n(n): # Logarithmic Time
    # Use a small value for n to avoid log(0) if n_values starts from 0 or has very small values.
    # In a real scenario, n is usually >= 1.
    return np.log2(n) if n > 0 else 0

def O_n(n): # Linear Time
    return n

def O_n_log_n(n): # Log-Linear Time
    return n * np.log2(n) if n > 0 else 0

def O_n_squared(n): # Polynomial Time (n^2)
    return n**2

def O_2_pow_n(n): # Exponential Time (2^n)
    return 2**n

# --- Generate data points ---

# Choose a range for n. For exponential, it grows very fast, so keep it relatively small.
n_values = np.arange(1, 51) # From 1 to 50 for better visualization of all curves

time_1 = [O_1(n) for n in n_values]
time_log_n = [O_log_n(n) for n in n_values]
time_n = [O_n(n) for n in n_values]
time_n_log_n = [O_n_log_n(n) for n in n_values]
time_n_squared = [O_n_squared(n) for n in n_values]
time_2_pow_n = [O_2_pow_n(n) for n in n_values]

# --- Plotting ---

plt.figure(figsize=(12, 8))

plt.plot(n_values, time_1, label='O(1) - Constant', linestyle='-')
plt.plot(n_values, time_log_n, label='O(log n) - Logarithmic', linestyle='-')
plt.plot(n_values, time_n, label='O(n) - Linear', linestyle='-')
plt.plot(n_values, time_n_log_n, label='O(n log n) - Log-Linear', linestyle='-')
plt.plot(n_values, time_n_squared, label='O(n^2) - Polynomial', linestyle='-')
# Exponential grows too fast to plot on the same scale for larger n
# We'll plot it for a very small subset of n to show its steepness
plt.plot(n_values[:10], time_2_pow_n[:10], label='O(2^n) - Exponential', linestyle='-', marker='o')

plt.xlabel("Input Size (n)")
plt.ylabel("Operations / Time")
plt.title("Comparison of Time Complexities")
plt.legend()
plt.grid(True)
plt.ylim(0, 1000) # Limit y-axis for better visibility of lower complexities
plt.xlim(0, 50) # Set x-axis limit explicitly
plt.show()
```

```python
# Write a function to perform quick sort and show that it has Logarithmic Time Complexity O(log n)
```

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    else:
        pivot = arr[len(arr) // 2] # Choose a pivot element (middle element for simplicity)
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

# Example usage of Quick Sort
example_array = [3, 6, 8, 10, 1, 2, 1]
sorted_array = quick_sort(example_array)
print(f"Original array: {example_array}")
print(f"Sorted array: {sorted_array}")
```

This is a basic implementation of Quick Sort. To demonstrate its average-case O(n log n) complexity visually, we can plot the `n * log(n)` function, similar to how we plotted Merge Sort earlier, as both algorithms share this average-case complexity.

```python
import math
import matplotlib.pyplot as plt
import numpy as np

# Function to simulate O(n log n) time complexity
def O_n_log_n_simulated(n):
    if n <= 1:
        return 1
    return n * np.log2(n)

# Define a range of input sizes
n_values = np.arange(1, 1001)

# Calculate simulated time values
time_values_n_log_n = [O_n_log_n_simulated(n) for n in n_values]

# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(n_values, time_values_n_log_n, label='O(n log n) - Simulated Quick Sort/Merge Sort')
plt.xlabel("Input Size (n)")
plt.ylabel("Operations / Time")
plt.title("O(n log n) Log-Linear Time Complexity Simulation")
plt.grid(True)
plt.legend()
plt.show()
```
